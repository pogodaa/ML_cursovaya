def component_contribution_analysis(decomposition):
    """Анализ вклада каждой компоненты в общую дисперсию"""
    
    # Вычисляем дисперсии компонент
    observed_var = decomposition.observed.var()
    trend_var = decomposition.trend.dropna().var() if decomposition.trend.dropna().size > 0 else 0
    seasonal_var = decomposition.seasonal.dropna().var() if decomposition.seasonal.dropna().size > 0 else 0
    residual_var = decomposition.resid.dropna().var() if decomposition.resid.dropna().size > 0 else 0
    
    print("📈 ВКЛАД КОМПОНЕНТ В ОБЩУЮ ДИСПЕРСИЮ:")
    print("-" * 55)
    print(f"{'Компонента':<15} {'Дисперсия':<12} {'Доля':<10} {'Интерпретация'}")
    print("-" * 55)
    
    components = [
        ('Тренд', trend_var),
        ('Сезонность', seasonal_var), 
        ('Остатки', residual_var)
    ]
    
    for name, var in components:
        if observed_var > 0:
            proportion = var / observed_var
            if proportion < 0.1:
                interpretation = "Слабый вклад"
            elif proportion < 0.3:
                interpretation = "Умеренный вклад"
            elif proportion < 0.6:
                interpretation = "Сильный вклад"
            else:
                interpretation = "Доминирующий вклад"
            
            print(f"{name:<15} {var:>10.6f} {proportion:>9.1%} {interpretation:>15}")
    
    # Объясненная дисперсия
    explained_variance = (trend_var + seasonal_var) / observed_var if observed_var > 0 else 0
    print(f"\n🎯 ОБЪЯСНЕННАЯ ДИСПЕРСИЯ: {explained_variance:.1%}")
    print(f"🎲 НЕОБЪЯСНЕННАЯ ДИСПЕРСИЯ: {1 - explained_variance:.1%}")

from scipy.signal import periodogram

def spectral_analysis(df, target_col='Global_active_power'):
    # Берем дневные данные
    daily_data = df[target_col].resample('D').mean().dropna()
    
    # Вычисляем периодограмму
    frequencies, power_density = periodogram(daily_data, fs=1.0)  # fs=1 для дневных данных
    
    # Преобразуем частоты в периоды (в днях)
    periods = 1 / frequencies[1:]  # исключаем нулевую частоту
    power_density = power_density[1:]
    
    # Находим наиболее значимые периоды
    significant_periods = periods[power_density > np.percentile(power_density, 90)]
    
    plt.figure(figsize=(12, 6))
    plt.semilogy(periods, power_density)
    plt.title('Периодограмма временного ряда')
    plt.xlabel('Период (дни)')
    plt.ylabel('Спектральная плотность мощности')
    plt.grid(True, alpha=0.3)
    
    # Подписываем значимые периоды
    for period in significant_periods[:5]:  # Топ-5 значимых периодов
        if 1 <= period <= 365:  # Ограничиваем разумными периодами
            plt.axvline(x=period, color='red', linestyle='--', alpha=0.7)
            plt.text(period, np.max(power_density)/2, f'{period:.1f} дн', 
                    rotation=90, va='bottom')
    
    plt.show()
    
    print("Наиболее значимые периоды:")
    for period in sorted(significant_periods[:5]):
        if 1 <= period <= 365:
            print(f"  {period:.1f} дней")

        if var != target_var:
            pearson_val = pearson_corr.loc[target_var, var]
            spearman_val = spearman_corr.loc[target_var, var]
            difference = spearman_val - pearson_val
            abs_diff = abs(difference)

def statistical_noise_analysis(residuals):
    """Статистический анализ свойств шумов"""
    
    from scipy.stats import shapiro, normaltest, jarque_bera
    
    print("📊 ТЕСТЫ НОРМАЛЬНОСТИ РАСПРЕДЕЛЕНИЯ:")
    
    # Тест Шапиро-Уилка (для n < 5000)
    if len(residuals) <= 5000:
        shapiro_stat, shapiro_p = shapiro(residuals)
        print(f"• Тест Шапиро-Уилка: p-value = {shapiro_p:.6f}")
        print(f"  {'✅ Нормальное' if shapiro_p > 0.05 else '❌ Не нормальное'} распределение")
    
    # Тест на нормальность Д'Агостино
    dagostino_stat, dagostino_p = normaltest(residuals)
    print(f"• Тест Д'Агостино: p-value = {dagostino_p:.6f}")
    print(f"  {'✅ Нормальное' if dagostino_p > 0.05 else '❌ Не нормальное'} распределение")
    
    # Тест Харке-Бера
    jb_stat, jb_p = jarque_bera(residuals)
    print(f"• Тест Харке-Бера: p-value = {jb_p:.6f}")
    print(f"  {'✅ Нормальное' if jb_p > 0.05 else '❌ Не нормальное'} распределение")
    
    print(f"\n📊 АСИММЕТРИЯ И ЭКСЦЕСС:")
    print(f"• Асимметрия (skewness): {residuals.skew():.4f}")
    print(f"  {'✅ Симметричное' if abs(residuals.skew()) < 0.5 else '⚠️ Умеренно асимметричное' if abs(residuals.skew()) < 1 else '❌ Сильно асимметричное'}")
    print(f"• Эксцесс (kurtosis): {residuals.kurtosis():.4f}")
    print(f"  {'✅ Нормальный эксцесс' if abs(residuals.kurtosis()) < 0.5 else '⚠️ Умеренный эксцесс' if abs(residuals.kurtosis()) < 1 else '❌ Высокий эксцесс'}")
    
    # Анализ автокорреляции
    print(f"\n🔍 АНАЛИЗ АВТОКОРРЕЛЯЦИИ ШУМОВ:")
    lags_to_check = [1, 7, 14, 30]
    acf_vals = acf(residuals, nlags=30)
    
    significant_autocorr = False
    for lag in lags_to_check:
        if lag < len(acf_vals):
            corr_val = acf_vals[lag]
            if abs(corr_val) > 1.96 / np.sqrt(len(residuals)):  # 95% доверительный интервал
                print(f"• Lag {lag}: {corr_val:.4f} ❌ ЗНАЧИМАЯ автокорреляция")
                significant_autocorr = True
            else:
                print(f"• Lag {lag}: {corr_val:.4f} ✅ НЕ значимая")
    
    if not significant_autocorr:
        print("✅ Шумы можно считать 'белым шумом' - нет значимой автокорреляции")

ВВЕДЕНИЕ

Вода - ключевой компонент для существования и развития любого современного города. Внешние условия, такие как: урбанизация, изменение погодных условий, износ инфраструктуры и непредвиденные обстоятельства (например, аварии на магистралях), создают сложные и динамичные условия эксплуатации городских систем водоснабжения. Эти факторы напрямую влияют на объем потребления воды и могут вызывать непредвиденно большие расходы столь ценного ресурса, а также приводить к дисбалансу между спросом и предложением, что чревато рисками для устойчивого развития городской среды.
В условиях такой нестабильности традиционные методы прогнозирования, основанные на усредненных исторических данных, становятся недостаточно эффективными. Они не способны оперативно учитывать влияние множества внешних и внутренних факторов, что чревато значительными рисками для устойчивости и экономической эффективности городской среды.
В данной работе обосновывается разработка приложения для прогнозирования водопотребления как инструмента финансового менеджмента и управления рисками. Точный прогноз позволяет оптимизировать закупки реагентов и электроэнергии, планировать ремонтные кампании и избегать штрафов за несоответствие качества услуг. Таким образом, предлагаемое решение позиционируется не просто как IT-продукт, а как стратегическая система поддержки принятия управленческих решений, направленная на повышение финансовой устойчивости предприятия водоснабжения и городского бюджета в целом. 
Ожидаемым экономическим эффектом от внедрения системы является не только прямая экономия средств на электроэнергию и реагенты (оцениваемая в 5-15% от соответствующих статей бюджета), но и существенное снижение непредвиденных расходов. Это выражается в сокращении бюджета на ликвидацию аварий, уменьшении объема списанной неучтенной воды и предотвращении штрафных санкций. Таким образом, инвестиции в разработку приложения окупаются за счет перевода значительной части условно-переменных и непредвиденных расходов в категорию планируемых и управляемых затрат, что является краеугольным камнем финансовой стабильности и основой для долгосрочного стратегического планирования развития городской инфраструктуры.


1 АНАЛИЗ ПРЕДМЕТНОЙ ОБЛАСТИ

Цель разработки - разработать web-приложение с понятным для пользователя интерфейсом которое сможет эффективно предсказывать будущее потребление воды на основе загруженных в него данных. 
Данная работа посвящена комплексному анализу данных и обучению прогнозных моделей на их основе. Полученные модели будут интегрированы в приложение, разработанное на языке Python. Такой подход позволит не только продемонстрировать эффективность моделей, но и создать основу для практического инструмента, который может быть использован городскими службами для планирования и оперативного управления ресурсами.
Ключевым этапом при выполнении работы является тщательный сбор данных из разнородных источников, их консолидация в единый датасет и всесторонняя подготовка к дальнейшему анализу. Этот этап включает в себя очистку данных от аномалий и шума, обработку пропущенных значений, а также генерацию новых признаков (feature engineering), которые могут повысить предсказательную силу моделей. Например, помимо исторических данных о потреблении воды, в анализ могут быть включены такие параметры, как календарные атрибуты (день недели, праздники), метеорологические данные (температура, осадки) и экономические индикаторы.  
Одна из задач разработки приложения - сравнительный анализ нескольких типов прогнозных моделей машинного обучения. Это необходимо для расширения функционала приложения и для получения наиболее точных и надежных результатов предсказания потребления воды как в краткосрочной (оперативное планирование на сутки или неделю вперед), так и в долгосрочной перспективе (стратегическое планирование на сезон или год). Планируется испытать как классические линейные модели (например, ARIMA, регрессионный анализ), так и более сложные ансамблевые алгоритмы (такие как Gradient Boosting) и, возможно, простые рекуррентные нейронные сети (RNN), демонстрирующие высокую эффективность при работе с временными рядами. 
Исполнение курсовой работы требует комплексного применения теоретических знаний и практических навыков, приобретенных в ходе изучения курса машинного обучения и анализа данных. Для начала крайне важно детально рассмотреть и формализовать проблематику темы. Это подразумевает изучение специфики управления водными ресурсами в условиях современного мегаполиса, идентификацию ключевых факторов, влияющих на потребление, и анализ потенциальных рисков, связанных с неточным прогнозированием. Только глубокое понимание предметной области позволит корректно подойти к постановке задачи машинного обучения - определить целевую переменную (например, суточный объем потребления воды в кубометрах), выбрать релевантные признаки и сформулировать критерии оценки качества моделей (метрики типа MAE, RMSE, MAPE).
Для задач подобного уровня понадобятся модели машинного обучения, которые хорошо чувствуют тренд, сезонность и стабильно работают с разными данными. Для этого хорошо подойдут модели ARIMA, SARIMA, CatBoost, т.к. предстоящая задача - это логистическая регрессия. Одна из главных проблем состоит в том, что модели, которые хорошо предсказывают на короткий срок времени оказываются малоэффективны для предсказания в долгосрочной перспективе и наоборот.
ARIMA (Autoregressive Integrated Moving Average) - Стандарт для прогнозирования стационарных временных рядов. Как понятно из названия, модель сочетает в себе авторегрессию, интегрирование и скользящее среднее. Ключевое примущество модели - высокая эффективность работы с хорошо структурированными данными. При этом мультисезонность и учёт внешних факторов у этой модели может быть не самым лучшим.
SARIMA (Seasonal ARIMA) - Улучшенная версия модели ARIMA которая имеет преимущество в учитывании сезонных компонентов. Для данных о потреблении воды в разные периоды времени модель SARIMA является привлекательным кандидатом.
CatBoost (Categorcial Boosting) - Высокая точность, скорость обучения, минимальная предобработка данных, интерпретируемость и устойчивость к переобучению. Все эти признаки содержит кошачья модель, на неё возложены высокие ожидания при работе с данными из-за специфики её работы с временными рядами и гиперпараметрами. 
Следовательно, необходимо применить ансамблевый метод и разработать конвейерную систему моделей, которая будет анализировать входящие данные, фильтровать аномалии и рассматривать разные предсказания. После этого осуществляется динамический выбор наиболее точного предсказания на период времени от всех моделей. 
Множество моделей, которые будут применяться в ходе решения задачи подталкивают на использования метода стекинга (Stacking) - создание одной метамодели, которая строит финальное предсказание на основе предсказаний, которые произвели использованные модели. Это поможет создать наиболее оптимальное и точное решение, которое будет удовлетворять необходимые запросы пользователей.
Ожидаемый результат от web-приложения - это аналитический инструмент мониторинга, который будет предсказывать будущее потребление воды,  на основе чего рядовой пользователь сможет изучить потребление воды в временном промежутке и по необходимости регулировать своё потребление воды. Если же речь идёт о компании, частной или управляющей, то отследить потребление воды и экономические особенности, что в свою очередь является важным фактором в обслуживании водопроводных сетей и управлением собственными ресурсами.
Для того чтобы понять проблематику, которая стоит перед разработкой приложения необходимо провести комплексный анализ проблем, которые возникают в современных системах водоснабжения.
1.1 Техническое состояние инфраструктуры
Очень часто одной из технических проблем является корреляция реального уровня потребления воды с её подачей. Погодный условия такие как жара или холод напрямую влияют на потребление воды в городе, недостаточная подача воды означает падение давления в сетях водоснабжения, что вызывает сбои в подаче воды и сбои в работе предприятий. Отсутствие прогнозирования с учётом рисков влечёт за собой колоссальный финансовые потери.
Система водоснабжения в любом городе содержит один из самых важных факторов - техническое состояние. Любые аварии (а без надлежащего обслуживания они будут происходить регулярно) приводят к сбоям в работе предприятий и к проблемам с водоснабжением во всём городе.
1.2 Нестабильность факторов влияния
Предсказание потребления воды - это сложный процесс, который напрямую зависит от факторов влияния, которые в свою очередь крайне нестабильны. Погодные условия, сезон, социальные условия, экономическая ситуация и другие факторы могут влиять на потребление воды. Крайне важно их учитывать это при изучении полученной базы данных и построении моделей.
Сфера водоснабжения требует прогнозов на разные временные и сезонные промежутки, это означает необходимость разработки таких моделей машинного обучения, которые могут функционировать в разных условиях прогнозирования.
Для дежурных служб необходим точный краткосрочный почасовой прогноз по дням, потребление воды в утренние и вечерние периоды, потребление воды в будние дни и выходные. Если речь идёт о руководстве планово-экономических отделов, то необходим долгосрочный прогноз на месяц, полгода или даже несколько лет. 
1.3 Многошаговое сезонное прогнозирование
Многошаговое сезонное прогнозирование (Multi-step forecasting) - это задача предсказания временного ряда на несколько шагов вперёд. Для выполняемой работы многошаговое сезонное прогнозирование это предсказание объёмов потребления на несколько временных интервалов вперёд - дней, недель, месяцев или лет. 
Реализация этого подхода потребует использования лагов кратных сезонности и циклических признаков времени. Это необходимо для наиболее точного предсказания используемых моделей. Комбинирование моделей SARIMA и CatBoost позволит достичь высокой точности на различных временных уровнях прогнозирования. При этом крайне важно правильно выделить сезонные компоненты в разной периодичности. 
В качестве конкурентов выбраны основные категории имеющегося ПО, выявляются их преимущества и недостатки на фоне выбранного решения.

3 Охрана труда и информационная безопасность
Разработка или использование web-приложения мало связана с производственными рисками и физическими травмами сотрудников, но необходимо учитывать эргономику рабочего места и создать оптимальные условия рабочего места пользователя.
Рабочее место должно быть оборудовано в соответствии с санитарно-техническими нормами: необходимое освещение, хорошая вентиляция, звукоизоляция. Монитор должен располагаться на расстоянии 50-70 см от глаз пользователя (значение может отличаться из-за индивидуальных особенностей человека), расположения монитора должно располагаться относительно источников света таким образом, чтобы не было бликов, сам же монитор должен быть оборудован антибликовыми фильтрами, если пользователь имеет проблемы с зрением, то ему необходимо носить очки с специальным напылением, в некоторых случаях контактные линзы не могут оказать должного воздействия. Кресло должно быть регулируемым, с поддержкой поясничного и шейного отделов позвоночника, позволяющими снизить нагрузку на организм и сохранить правильную осанку. Клавиатура и мышь должны располагаться так, чтобы руки находились в нейтральном положении. 
Рекомендуется соблюдать режим работы и отдыха, каждый час необходимо делать перерывы по 5-10 минут и проводить гимнастику глаз, шеи, плеч, спины и рук. Это поможет держать тело в тонусе и держать производственную продуктивность на высоком уровне. 
Вспоминаем про безопасность работы с электрооборудованием - получить разряд электричества или оказаться посреди пожара не хочет никто. Вся электроника в округе (периферия, серверы, рабочие места и т.д.) должны быть заземлены и находиться в исправном состоянии. Если системный блок, блок питания, электрощиток или другое оборудование находится под напряжением - вскрывать их запрещается, необходимо принять действия прописанные в технике безопасности при работе с электрооборудованием. Рабочие места должны быть оснащены средствами пожаротушения (порошковый или углекислотный огнетушитель), загромождение проходов к пожарным щиткам и аварийным выходам, а также использование легковоспламеняемых материалов и воды (вода и электричество нежелательное для пользователя сочетание) строго запрещено. Регулярные проверки, правильная подготовка сотрудников и тестирование оборудования сильно уменьшит риск потенциальных аварийных ситуаций.
