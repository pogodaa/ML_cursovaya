Обоснование актуальности выбранной темы

Освещение актуальности не должно быть многословным. В пределах одной страницы необходимо показать суть проблемной ситуации, а также имеющееся противоречие, из чего можно будет сделать вывод об актуальности темы. Правильная постановка проблемы имеет очень большое значение в методологии. Обоснование проблемы определяет стратегию выполнения курсовой работы (проекта) и направления поиска ее анализа. 
Определение объекта и предмета 
Постановка цели предполагает определение объекта и предмета проектной работы. 
Объект — это явление (процесс), порождающее проблемную ситуацию и взятое для изучения. 
Затем выделяется предмет курсовой работы (проекта) — это тот угол зрения, под которым смотрят на объект работы. 
Предмет — это то, что находится в границах объекта, та часть объекта работы, которая подлежит специальному изучению и анализу. 
Объект и предмет курсовой работы (проекта) как категории процесса работы над курсовым проектом соотносятся между собой как общее и частное. Для облегчения восприятия данных категорий следует помнить, что объект работы — это некий процесс или явление, а предмет — это содержательная часть этого процесса или явления. 
Следует помнить важное правило: предмет всегда должен находить отражение в теме работы, но не дублировать ее и цель курсовой работы (проекта). 
Постановка цели и задач исследования 
Цель курсовой работы (проекта) — это словесно — логическое описание представления о результате курсовой работы (проекта), того, что ожидается в итоге проведенной работы. Входит в научный аппарат исследования и в курсовой работе формулируется на основе аналитической деятельности. 
Очень важным является тот факт, что цель всегда должна находить отражение в теме курсовой работы (проекта).
Далее указываются конкретные задачи, которые предстоит решить для достижения указанной цели. Обычно это делается в форме перечисления (изучить, обобщить, выяснить, проанализировать). 
Корректная формулировка задач является очень важным моментом при написании курсовой работы (проекта), так как описание их решения составляет содержание курсовой работы (проекта). Важность данного процесса определяется тем, что заголовки глав и параграфов исходят из формулировок задач. 
Рассмотрим конкретный пример. 
Тема: «Разработка системы защиты персональных данных в МУП ЖКХ г. Межгорье». 
Целью работы является разработка системы защиты персональных данных в МУП ЖКХ г. Межгорье. 
Для выполнения цели работы необходимо определить ряд сопутствующих задач, а именно:
– проанализировать объект защиты, его род деятельности;
– определить защищаемую информацию;
– составить модель угроз;
– сформировать рекомендации по защите информации от актуальных угроз;
– составить организационно-распорядительную документацию.
Объектом исследования в данной работе является система защиты персональных данных на предприятии МУП ЖКХ г. Межгорье. Предметом исследования является разработка системы защиты персональных данных в МУП ЖКХ г. Межгорье.



Практическая значимость. 
Практическая значимость заключается в возможности использования его результатов в практической деятельности, независимо от того, какой характер носит курсовая работа. 
Структура курсовой работы (проекта). 
Пример: «Курсовая работа изложена на 35 страницах машинописного текста, состоит из введения, трех глав, заключения, списка использованных источников и приложения. В работе представлено 7 рисунков и 4 таблицы». 
Необходимо отметить важное правило — введение, как и заключение, рекомендуется писать после полного завершения основной части. До того, как будет написана основная часть работы, реально невозможно написать хорошее введение, так как автор еще не вполне овладел материалами по теме. 
Первая глава: теоретическая часть. 
Первая глава содержит изложение теоретических основ работы, обзор концепций по рассматриваемой проблеме, описание сущности анализируемого предмета, содержания основных категорий и понятий. 
В этом разделе, представляющем теоретическую часть курсовой работы (проекта), проводится обзор научных, научно-методических и иных источников информации, анализируются основные проблемы выбранной темы, отражаются мнения различных авторов, приводятся выводы обучающегося. Он должен указать и обосновать точку зрения, которой придерживается. 
Название этого раздела должно соответствовать выбранной теме, но не должно дублировать ее название. Содержание первой главы должно включать не менее 2 – х подразделов. 
Вторая глава: практическая часть. 
Первый пункт второй главы включает в себя анализ изучаемой проблемы. Здесь нельзя ограничиваться одной лишь констатацией фактов. Важно найти причины возникновения проблемы. 
В курсовой работе второй раздел может быть представлен практической частью, в которой содержатся расчеты, графики, таблицы, схемы и т.п.. 
Второй пункт второй главы: разработка рекомендаций и мероприятий по решению изучаемой проблемы на основе ее детального и всестороннего анализа. 
Каждая глава в курсовой работе должна заканчиваться выводом. Объём вывода обычно составляет 1–2 абзаца или треть страницы. 
Заключение включает в себя выводы по курсовой работе, сформулированные в ходе ее выполнения и содержит обобщенные результаты выполнения работы. Важно помнить, что выводы ОБЯЗАТЕЛЬНО должны соотноситься с задачами работы, то есть ВЫВОД — ЭТО КОНКРЕТНЫЙ ОТВЕТ НА РЕШЕНИЕ ОПРЕДЕЛЕННОЙ ЗАДАЧИ. Соответственно, количество выводов не должно быть меньше или больше количества задач курсовой работы (проекта). 
Список использованных источников является структурной частью любой работы. Он помещается после основного текста работы и позволяет автору документально подтвердить достоверность и точность приводимых в тексте заимствований: цитат, идей, фактов, таблиц, иллюстраций, формул и других документов, на основе которых строится работа. Каждый документ, включенный в список использованных источников, должен быть описан в соответствии с требованиями ГОСТ Р 7.0.5 – 2008 Библиографическая ссылка. Общие требования и правила составления. Оформление сносок и ссылок 
Если необходимо пояснить отдельные данные, приведенные в пояснительной записке, 
то эти данные следует обозначать надстрочными знаками сноски. Знак сноски выполняют 
арабскими цифрами со скобкой непосредственно после того слова, числа, символа, 
предложения, к которому дается пояснение. 
Пример: 
«..печатающее устройство 2)» 
Нумерация сносок отдельна для каждой страницы. Допускается вместо цифр применять 
звездочки *, но не более 4-х. В нижней части страницы проводят короткую тонкую 
горизонтальную линию с левой стороны. Под ней с абзацного отступа указывают знак сноски 
и дают текст пояснения. 
В пояснительной записке допускаются ссылки на данный документ, стандарты и другие 
документы. Ссылаться следует на документ в целом или его разделы и приложения. Не 
допускается ссылаться на подразделы, пункты, таблицы другого документа. 
Ссылки в тексте на формулы данного документа приводятся в круглых скобках, 
например, «... в формуле (1)». Ссылки на источники, указанные в списке литературы, 
приводятся в квадратных скобках. Например, «Как указано в исследовании П.О. Петрова [12], 
...»

ВВЕДЕНИЕ
 
Актуальность работы. С каждым годом увеличивается количество различных транспортных средств, как в личной собственности, так и в собственности предприятий. Рост парка транспорта создаёт запрос создать маркетплейс для всех видов транспорта. Итак, маркетплейсы по продаже транспорта представляют собой современные и инновационные платформы, которые революционизируют способы покупки, продажи и обмена транспортными средствами. Эти онлайн-ресурсы создают уникальное пространство для владельцев ТС., автодилеров и покупателей, предоставляя им возможность находить, выбирать и приобретать транспортные средства наиболее удобным и выгодным способом.
Целью настоящего исследования является анализ и обсуждение значимости и функциональности маркетплейсов по продаже транспорта в современном мире. Мы рассмотрим основные преимущества и задачи, решаемые такими платформами, а также их влияние на рынок транспорта и сферу автодилерства.
Значимость маркетплейсов по продаже транспорта:
1. Расширение выбора и доступности. Маркетплейсы транспорта предоставляют покупателям широкий спектр вариантов автомобилей, самолетов и других видов т. с. включая новые, подержанные и редкие модели. Это увеличивает возможности выбора и удовлетворяет разнообразные потребности клиентов.
2. Удобство и эффективность. Покупка или продажа т. с. через маркетплейс может быть выполнена онлайн, без необходимости посещать автосалоны. Это экономит время и упрощает процесс сделки.
3. Прозрачность и доверие. Маркетплейсы обеспечивают подробные описания транспорта, историю обслуживания, отзывы от предыдущих владельцев и дилеров. Это помогает покупателям принимать информированные решения и увеличивает доверие к продавцам.
4. Снижение посредников. Маркетплейсы минимизируют роль посредников в сделках, что позволяет продавцам и покупателям взаимодействовать напрямую и избегать дополнительных комиссий.
5. Инновации и развитие. Конкуренция на маркетплейсах стимулирует автодилеров и продавцов предлагать лучшие условия и новые технологии, что способствует развитию автомобильной индустрии.
Маркетплейс – прибыльное предприятие, а финансовые вложения в маркетплейс на первом этапе минимальны и равны покупке сервера и разработке прототипа маркетплейса. На рынке аналога чисто по всем видам транспортных средств нет. Спрос обуславливает предложение, автомобилестроение развивается, все люди хотят покупать электросамокаты и другие виды транспорта соответственно можно сделать одну площадку для продажи любого типа и вида ТС. 
Непосредственно вопросы разработки системы маркетплейса транспорта, пользующиеся маркетплейсами по продаже автотранспорта, водного и летательного транспорта является предметами исследования в настоящей работе.

Рисунок 1 – Пример для разработки системы автоматизации

В данной информационной системе предусмотрена функциональность выбора категории необходимого товара. Того, который необходим покупателю. Так же выбор города, для поиска объявлений поблизости. 
	В работе данной информационной системы в основном участвуют продающие лица. Они заполняют информацию о своем товаре (предоставляют фото, описание, характеристики и цену), чтобы потенциальный покупатель имел максимум информации о товаре, который он в данный момент предоставляет. Так же продавец предоставляет свой номер телефона, по которому потенциальный покупатель сможет связаться с ним
Задача информационной системы – облегчение процесса поиска клиентов продавцам, облегчение поиска нужного товара покупателю.
Таким образом, предметом автоматизации является создание информационной системы, позволяющей легко продавцам выставлять свои ТС на продажу, а также позволяющей покупателям иметь полное представление о товаре, для облегчения поиска нужного транспорта.
Проведенный анализ предметной области позволяет сформулировать задачу на проектирование:
- разработать информационную систему автоматизации для продажи транспорта в сети Интернет.
- реализовать рекомендательную систему | поиск с помощью алгоритмов МЛ (для удобства пользователя).
Решение задачи с использованием современных информационных технологий представляется возможным.

можно разделить на подзадачи, требующие конкретной реализации.

	- функциональность авторизации / аутентификации в информационную систему
	- спроектировать базу данных для хранения информации о пользователях, а также о объявлениях, размещенных пользователями в данной информационной системе
	- разработать backend часть, при помощи которой будет проходить работа с разработанной базой данных, а также предоставлена возможность взаимодействия с frontend частью
	- разработать единую дизайн-систему для сохранения единого дизайн-стиля на сайте информационной системы
	- разработать сайт, адаптированный под любые размеры устройств, для отображения доступных объявлений по выбранной категории, предоставляющий функциональность создания объявлений, их редактирование, а также удаление объявления из информационной системы
	- разработать приложение под Android и iOS системы, предоставляющий тот же функционал, что и сайт (при достаточном количестве времени, приложение WEB 3.0)
	- сделать презентационные материалы для презентации данной информационной системы.

Проектировка базы данных
Проектирование базы данных – это важный этап при создании системы хранения и управления данными. Оно включает в себя несколько ключевых этапов:
Определение требований: начать с определения целей базы данных, типов данных, которые нужно хранить, и их взаимосвязей. Определить, какие данные будут храниться и как они будут использоваться.
Создание концептуальной модели: разработать концептуальную модель, которая отражает общую структуру базы данных без уточнения технических деталей. Использовать сущности, атрибуты и их взаимосвязи.
Преобразование в логическую модель: преобразовать концептуальную модель в логическую, используя конкретные технические средства (например, ER-диаграммы, UML). Это включает определение таблиц, полей, связей между таблицами и ограничений.
Нормализация данных: провести процесс нормализации, чтобы уменьшить избыточность данных и повысить их целостность.
Физическое проектирование: определить, как база данных будет физически реализована (типы данных, индексы, оптимизация запросов) с учетом требований к производительности.
Тестирование и оптимизация: после реализации базы данных провести тестирование для проверки ее работоспособности, целостности и производительности. Оптимизируйте структуру и запросы при необходимости.
Развитие и поддержка: после внедрения базы данных продолжать поддерживать ее, внося изменения и улучшения в соответствии с потребностями пользователя и бизнеса.
Хорошее проектирование базы данных требует внимательного анализа и понимания бизнес-потребностей, чтобы обеспечить эффективное хранение, доступ и управление данными. Внимательный анализ бизнес-потребностей – это ключевой аспект проектирования базы данных. Понимание того, какие данные необходимы для бизнес-процессов, как они связаны между собой и как они будут использоваться, позволяет создать базу данных, которая эффективно подходят эти потребности.

Написание логики регистрации и входа в ИС:
Функционал авторизации и регистрации довольно прост в реализации: при создании аккаунта мы сохраняем все значения полей из формы и отправляем эти значения в базу данных POST запросом из формы, который предназначен для создания сущности в базе данных. Если запрос прошел успешно – пользователя перенаправляет на основную страницу сайта, а его пароль сохраняется в браузере пользователя. Если же произошла ошибка – пользователь останется на странице регистрации
Функционал входа в аккаунт осуществляется следующим образом: при заполнении полей email и пароль и нажатии кнопки «Вход» в базу данных отправляется запрос на уже существующих пользователей. Если данный email уже есть в базе данных, то проверяется пароль пользователя, который достается из хранилища браузера (localStorage). Если все проверки пройдены успешно, то пользователя направляет на основную страницу сайта. Если же происходит ошибка (аккаунт был не найден, или пароль неверен), то пользователь будет получать соответствующее оповещение:

```python
Так теперь у меня есть пример курсовой которая уже прошла защиту на 5. ВОт эта работа, нужно запомнить. Приложение не прикладываю, но там одно приложение в котором просто код всей программы.



Министерство образования Московской области
Государственное бюджетное профессиональное образовательное учреждение Московской области «Физико-технический колледж»


                                                                           ОЦЕНКА_________________
                                                                           Руководитель специальности
                                                                           09.02.07 Информационные
                                                                           Системы и программирование
                                                                           __________________________

«Разработка приложения по предсказанию воды в городе»

КУРСОВАЯ РАБОТА ПО ПРЕДМЕТУ:
МДК 05.01 «Машинное обучение и Анализ данных»



Руководитель КП                                           Разработал
преподаватель                                                студент группы ИСП-23В
ГБПОУ МО «Физтех-колледж»                   Филиппов Д.М.
Коновалов  И.В.                                             
   





Долгопрудный 2025


СОДЕРЖАНИЕ

1 АНАЛИЗ ПРЕДМЕТНОЙ ОБЛАСТИ	6
1.1 Техническое состояние инфраструктуры	9
1.2 Нестабильность факторов влияния	9
1.3 Многошаговое сезонное прогнозирование	10
ВВЕДЕНИЕ	4
2 ПРАКТИЧЕСКАЯ ЧАСТЬ. РАЗРАБОТКА И РЕАЛИЗАЦИЯ	15
2.1 Осмотр и анализ данных	15
2.2 Построение и обучение моделей	23
2.3 Разработка и тестирование приложения	29
3 Охрана труда и информационная безопасность	32
3.1 Информационная безопасность	33
ЗАКЛЮЧЕНИЕ	35
СПИСОК ИСПОЛЬЗОВАННЫХ ИСТОЧНИКОВ	37
ПРИЛОЖЕНИЯ	39


ВВЕДЕНИЕ

Вода - ключевой компонент для существования и развития любого современного города. Внешние условия, такие как: урбанизация, изменение погодных условий, износ инфраструктуры и непредвиденные обстоятельства (например, аварии на магистралях), создают сложные и динамичные условия эксплуатации городских систем водоснабжения. Эти факторы напрямую влияют на объем потребления воды и могут вызывать непредвиденно большие расходы столь ценного ресурса, а также приводить к дисбалансу между спросом и предложением, что чревато рисками для устойчивого развития городской среды.
В условиях такой нестабильности традиционные методы прогнозирования, основанные на усредненных исторических данных, становятся недостаточно эффективными. Они не способны оперативно учитывать влияние множества внешних и внутренних факторов, что чревато значительными рисками для устойчивости и экономической эффективности городской среды.
В данной работе обосновывается разработка приложения для прогнозирования водопотребления как инструмента финансового менеджмента и управления рисками. Точный прогноз позволяет оптимизировать закупки реагентов и электроэнергии, планировать ремонтные кампании и избегать штрафов за несоответствие качества услуг. Таким образом, предлагаемое решение позиционируется не просто как IT-продукт, а как стратегическая система поддержки принятия управленческих решений, направленная на повышение финансовой устойчивости предприятия водоснабжения и городского бюджета в целом. 
Ожидаемым экономическим эффектом от внедрения системы является не только прямая экономия средств на электроэнергию и реагенты (оцениваемая в 5-15% от соответствующих статей бюджета), но и существенное снижение непредвиденных расходов. Это выражается в сокращении бюджета на ликвидацию аварий, уменьшении объема списанной неучтенной воды и предотвращении штрафных санкций. Таким образом, инвестиции в разработку приложения окупаются за счет перевода значительной части условно-переменных и непредвиденных расходов в категорию планируемых и управляемых затрат, что является краеугольным камнем финансовой стабильности и основой для долгосрочного стратегического планирования развития городской инфраструктуры.


1 АНАЛИЗ ПРЕДМЕТНОЙ ОБЛАСТИ

Цель разработки - разработать web-приложение с понятным для пользователя интерфейсом которое сможет эффективно предсказывать будущее потребление воды на основе загруженных в него данных. 
Данная работа посвящена комплексному анализу данных и обучению прогнозных моделей на их основе. Полученные модели будут интегрированы в приложение, разработанное на языке Python. Такой подход позволит не только продемонстрировать эффективность моделей, но и создать основу для практического инструмента, который может быть использован городскими службами для планирования и оперативного управления ресурсами.
Ключевым этапом при выполнении работы является тщательный сбор данных из разнородных источников, их консолидация в единый датасет и всесторонняя подготовка к дальнейшему анализу. Этот этап включает в себя очистку данных от аномалий и шума, обработку пропущенных значений, а также генерацию новых признаков (feature engineering), которые могут повысить предсказательную силу моделей. Например, помимо исторических данных о потреблении воды, в анализ могут быть включены такие параметры, как календарные атрибуты (день недели, праздники), метеорологические данные (температура, осадки) и экономические индикаторы.  
Одна из задач разработки приложения - сравнительный анализ нескольких типов прогнозных моделей машинного обучения. Это необходимо для расширения функционала приложения и для получения наиболее точных и надежных результатов предсказания потребления воды как в краткосрочной (оперативное планирование на сутки или неделю вперед), так и в долгосрочной перспективе (стратегическое планирование на сезон или год). Планируется испытать как классические линейные модели (например, ARIMA, регрессионный анализ), так и более сложные ансамблевые алгоритмы (такие как Gradient Boosting) и, возможно, простые рекуррентные нейронные сети (RNN), демонстрирующие высокую эффективность при работе с временными рядами. 
Исполнение курсовой работы требует комплексного применения теоретических знаний и практических навыков, приобретенных в ходе изучения курса машинного обучения и анализа данных. Для начала крайне важно детально рассмотреть и формализовать проблематику темы. Это подразумевает изучение специфики управления водными ресурсами в условиях современного мегаполиса, идентификацию ключевых факторов, влияющих на потребление, и анализ потенциальных рисков, связанных с неточным прогнозированием. Только глубокое понимание предметной области позволит корректно подойти к постановке задачи машинного обучения - определить целевую переменную (например, суточный объем потребления воды в кубометрах), выбрать релевантные признаки и сформулировать критерии оценки качества моделей (метрики типа MAE, RMSE, MAPE).
Для задач подобного уровня понадобятся модели машинного обучения, которые хорошо чувствуют тренд, сезонность и стабильно работают с разными данными. Для этого хорошо подойдут модели ARIMA, SARIMA, CatBoost, т.к. предстоящая задача - это логистическая регрессия. Одна из главных проблем состоит в том, что модели, которые хорошо предсказывают на короткий срок времени оказываются малоэффективны для предсказания в долгосрочной перспективе и наоборот.
ARIMA (Autoregressive Integrated Moving Average) - Стандарт для прогнозирования стационарных временных рядов. Как понятно из названия, модель сочетает в себе авторегрессию, интегрирование и скользящее среднее. Ключевое примущество модели - высокая эффективность работы с хорошо структурированными данными. При этом мультисезонность и учёт внешних факторов у этой модели может быть не самым лучшим.
SARIMA (Seasonal ARIMA) - Улучшенная версия модели ARIMA которая имеет преимущество в учитывании сезонных компонентов. Для данных о потреблении воды в разные периоды времени модель SARIMA является привлекательным кандидатом.
CatBoost (Categorcial Boosting) - Высокая точность, скорость обучения, минимальная предобработка данных, интерпретируемость и устойчивость к переобучению. Все эти признаки содержит кошачья модель, на неё возложены высокие ожидания при работе с данными из-за специфики её работы с временными рядами и гиперпараметрами. 
Следовательно, необходимо применить ансамблевый метод и разработать конвейерную систему моделей, которая будет анализировать входящие данные, фильтровать аномалии и рассматривать разные предсказания. После этого осуществляется динамический выбор наиболее точного предсказания на период времени от всех моделей. 
Множество моделей, которые будут применяться в ходе решения задачи подталкивают на использования метода стекинга (Stacking) - создание одной метамодели, которая строит финальное предсказание на основе предсказаний, которые произвели использованные модели. Это поможет создать наиболее оптимальное и точное решение, которое будет удовлетворять необходимые запросы пользователей.
Ожидаемый результат от web-приложения - это аналитический инструмент мониторинга, который будет предсказывать будущее потребление воды,  на основе чего рядовой пользователь сможет изучить потребление воды в временном промежутке и по необходимости регулировать своё потребление воды. Если же речь идёт о компании, частной или управляющей, то отследить потребление воды и экономические особенности, что в свою очередь является важным фактором в обслуживании водопроводных сетей и управлением собственными ресурсами.
Для того чтобы понять проблематику, которая стоит перед разработкой приложения необходимо провести комплексный анализ проблем, которые возникают в современных системах водоснабжения.
1.1 Техническое состояние инфраструктуры
Очень часто одной из технических проблем является корреляция реального уровня потребления воды с её подачей. Погодный условия такие как жара или холод напрямую влияют на потребление воды в городе, недостаточная подача воды означает падение давления в сетях водоснабжения, что вызывает сбои в подаче воды и сбои в работе предприятий. Отсутствие прогнозирования с учётом рисков влечёт за собой колоссальный финансовые потери.
Система водоснабжения в любом городе содержит один из самых важных факторов - техническое состояние. Любые аварии (а без надлежащего обслуживания они будут происходить регулярно) приводят к сбоям в работе предприятий и к проблемам с водоснабжением во всём городе.
1.2 Нестабильность факторов влияния
Предсказание потребления воды - это сложный процесс, который напрямую зависит от факторов влияния, которые в свою очередь крайне нестабильны. Погодные условия, сезон, социальные условия, экономическая ситуация и другие факторы могут влиять на потребление воды. Крайне важно их учитывать это при изучении полученной базы данных и построении моделей.
Сфера водоснабжения требует прогнозов на разные временные и сезонные промежутки, это означает необходимость разработки таких моделей машинного обучения, которые могут функционировать в разных условиях прогнозирования.
Для дежурных служб необходим точный краткосрочный почасовой прогноз по дням, потребление воды в утренние и вечерние периоды, потребление воды в будние дни и выходные. Если речь идёт о руководстве планово-экономических отделов, то необходим долгосрочный прогноз на месяц, полгода или даже несколько лет. 
1.3 Многошаговое сезонное прогнозирование
Многошаговое сезонное прогнозирование (Multi-step forecasting) - это задача предсказания временного ряда на несколько шагов вперёд. Для выполняемой работы многошаговое сезонное прогнозирование это предсказание объёмов потребления на несколько временных интервалов вперёд - дней, недель, месяцев или лет. 
Реализация этого подхода потребует использования лагов кратных сезонности и циклических признаков времени. Это необходимо для наиболее точного предсказания используемых моделей. Комбинирование моделей SARIMA и CatBoost позволит достичь высокой точности на различных временных уровнях прогнозирования. При этом крайне важно правильно выделить сезонные компоненты в разной периодичности. 
В качестве конкурентов выбраны основные категории имеющегося ПО, выявляются их преимущества и недостатки на фоне выбранного решения.
1.4 Анализ конкурентов
Выбранное при разработке решение необходимо рассмотреть в том числе с точки зрения преимуществ и недостатков на фоне существующих конкурентов. Это необходимо для понимания возможностей уже существующего ПО и выявления рыночных ниш. Каждый имеющийся продукт в различной степени создаёт определённую планку функциональных требований для различных задач, которые они выполняют. Современное программное обеспечение для управления водными ресурсами предлагает как крупные корпоративные платформы, так и более лёгкие и простые приложения, вплоть до маленьких локальных проектов 5с открытым исходным кодом, разработанных энтузиастами.
1. Крупные платформы для управления коммунальной инфраструктурой. 
Эти платформы являются многоуровневыми комплексами решений класса Industrial IoT, предназначенные для цифрового отображения коммунального сектора. Их архитектура построена по принципу сквозной интеграции оборудования, данных и бизнес-процессов.
Пример: Siemens MindSphere, GE Predix, IBM Maximo.
Ключевые возможности:
1. Единая платформа для мониторинга всей инфраструктуры водоснабжения с поддержкой множества телеметрических датчиков.
2. Встроенные функции прогнозирования на основе запатентованных алгоритмов, учитывающих специфику коммунальной сферы.
3. Интеграция с SCADA-системами и диспетчерскими системами с OPC UA и MQTT протоколами.
4. Реализация цифровых двойников для симуляции и моделирования работы распределительных сетей.
5. Инструменты визуализации и аналитики с поддержкой технологии дополненной реальности. 
Недостатки:
1. Избыточный функционал для рядового пользователя и решения узкоспециализированных задач прогнозирования.
2. Долгие сроки внедрения и реализации в зависимости от сложности инфраструктуры.
3. Очень высокая цена владения и обслуживания.
4. Привязка к системе производителя и ограниченная совместимость со сторонним ПО.
1. Специальное программное обеспечение для водных каналов.
Данные решения разработаны исключительно для нужд организаций водоснабжения и работают по принципу оптимизации эксплуатации распределительных сетей.
Пример: Aquacue, WaterGEMS, Infowater.
Ключевые возможности: 
1. Полноценное моделирование сетей водоснабжения с учётом топологии местности, характеристик труб и насосного оборудования.
2. Развитые алгоритмы по обнаружению утечек с высокой точностью до 10 метров.
3. Модули прогнозирования потребления с учётом данных демографии и сезонности.
4. Инструменты планирования ремонтных работ.
5. Возможность симуляции различных аварийных ситуаций.
Недостатки:
1. Работа на стационарных рабочих местах с ограниченной мобильностью.
2. Слабая адаптивность алгоритмов к меняющимся условиям.
3. Ограниченная возможность модификации системы под особые требования заказчиков.
4. Сложность в эксплуатации системы, персоналу необходимо специальное обучение.
         
         2. Слабая адаптивность алгоритмов к меняющимся условиям.
         3. Ограниченная возможность модификации системы под особые   требования заказчиков. 
        
        3. Исследовательские разработки
Это направление представляет собой прикладные исследования в области машинного обучения и научные проекты с целью предсказания прогнозирование потребления воды.
Ключевые возможности:
1. Реализация LSTM, GAN, Transformers и других алгоритмов машинного обучения.
2. Обоснованность используемых методик и подходов к решению задач.
3. Разработка собственных экспериментальных решений с уникальными особенностями.
Недостатки:
1. Практически полное отсутствие готовых продуктов пригодных для массовой промышленной эксплуатации.
2. Крайне высокая сложность внедрения в реальных условиях.
3. Ограниченная техническая поддержка или её отсутствие.
4. Проблемы с производительностью и масштабируемостью рабочей системы.
Выбранное мной решение отличается простотой эксплуатации, что позволяет использовать приложение рядовому пользователю, а также это не потребует особого обучения для сотрудников различных компаний. Комбинирование нескольких моделей позволит выстроить адаптивную систему, которая сможет эффективно строить предсказание вне зависимости от резких изменений данных. Также приложение будет иметь широкую возможность модификации под требования заказчика без изменения изначальной архитектуры (если заказчик не потребует коррекции архитектуры по необходимости). 
Анализ существующих конкурентов показал, что существует две условные категории подобного ПО: дорогостоящие корпоративные платформы и ограниченные локальные решения. Обе категории представляют собой противоположные грани, вызывающие разрыв в области усреднённого решения. Разработанное приложение исключает избыточные механизмы  крупных платформ, сохраняя при этом высокую адаптивность к изменяющимся условиям и модульную архитектуру, позволяющую расширять функционал. 

2 ПРАКТИЧЕСКАЯ ЧАСТЬ. РАЗРАБОТКА И РЕАЛИЗАЦИЯ

При разработке мобильного приложения необходимо выстроить чёткий план работы с ориентированием на временные рамки и продуктивность работы. После планирования были выделены шаги:
1. Осмотр и анализ данных.
2. Построение и обучение моделей.
3. Разработка приложения и его тестирование.

2.1 Осмотр и анализ данных
Создал датасет на основе существующих данных при помощи ИИ установив временные и линейные зависимости между признаками.  После - корректировал датасет из-за небольших аномалий, которые были устранены модой и заменой средним значением. В процессе сздания датасета я выделил 26 колонок которые являются решающими факторами в предсказании потребления воды:
date - дата.
year - год.
month - месяц.
day - день.
day_of_week - день недели.
is_weekend - является выходным днем (0-нет, 1-да).
is_holiday - является праздничным днем (0-нет, 1-да).
avg_temperature - средняя температура.
min_temperature - минимальная температура.
max_temperature - максимальная температура.
precipitation - осадки.
humidity - влажность.
wind_speed - скорость ветра.
population - население.
water_price - цена на воду.
tourism_index - индекс туризма.
industrial_water_use - промышленное водопользование.
water_restrictions - ограничения на воду (0-нет, 1-да).
uv_index - УФ-индекс.
number_of_tourists - количество туристов.
construction_index - строительный индекс.
economic_index - экономический индекс.
water_main_breaks - прорывы водопроводных магистралей.
reservoir_level - уровень (глубина) водохранилища.
pool_opening_season - сезон открытия бассейнов (0-нет, 1-да).
garden_watering_season - сезон полива садов (0-нет, 1-да).
smart_meter_coverage - охват умными счетчиками.
water_consumption_tonnes - потребление воды (в тоннах).
Для некоторых классов было принято решение присвоить бинарное значение 0 ил 1 для простоты ориентации в базе данных. 
После этого был построен график временного ряда и были выведены тренд, сезонность и остаток. 
Тренд - долгосрочное направление изменения потребления.
Сезонность - годовые повторяющиеся циклы.
Остаток - случайные колебания, которые не объяснены трендом и сезонностью.
На рисунке 2.1 представлен оригинальный временной ряд потребления воды:
Рис.2.1 – оригинальный ряд потребления воды
На рисунке 2.2 представлен временной тренд:
Рис.2.2 – временное изменение тренда
На рисунке 2.3 представлена сезонная составляющая:
Рис.2.3 – сезонная составляющая по годовым циклам
На рисунке 2.4 представлены случайные колебания данных, не связанные с трендом и сезонностью:
Рис.2.4. – шумы в наших данных
При более детальном изучении статистических данных выявилось, что средний тренд составляет - 206948 тонн воды, сезонность по амплитуде составляет - 26149 тонн и остатки - 19234 тонн.
После установления и утверждения этих трёх параметров была проведена автокорреляция данных, она представлена на рисунке 2.5:
Рис 2.5 – корреляционные взаимодействия
Автокорреляция показывает, что параметры имеют не только временную связь, но и линейную.
Дополнительно для данных была проведена корреляция Пирсона и Спирмена.
Корреляция Пирсона (Pearson Correlation) - измеряет линейную зависимость между двумя переменными. Показывает, насколько хорошо связь между переменными описывается прямой линией. (Рис.2.6)
Корреляция Спирмена (Spearman Correlation) - измеряет монотонную зависимость между двумя переменными. Покажает, насколько хорошо связь описывается монотонной функцией (возрастающей или убывающей). (Рис.2.7)
Рис.2.6. – Матрица Корреляции Пирсона
Рис.2.7. – Корреляция Спирмена
Корреляции Пирсона и Спирмена показали, что данные имеют нетолько линейные, но и монотонные зависимости. 
Выделение скользящего среднего представлено на рисунке 2.8, это последний шаг перед обучением моделей:
Рис.2.8. Скользящее среднее  и тренд
2.2 Построение и обучение моделей
Для построения моделей пришлось немного изменить структуру кода для сезонной корректировки и разброса значений от минимального к максимальному в рамках доверительного интервала предсказания. Специально для приложения в каждую модель были добавлены встроенные рекомендации по планированию, инфраструктуре и ресурсам. На рисунке 2.9 представлено предсказание модели ARIMA: 

Рис.2.9 Модель ARIMA и её остатки
На рисунке 2.10 представлено предсказание модели SARIMA:
Рис.2.10 Модель SARIMA и её остатки
Ниже на рисунках (2.11- 2.16) представлены данные, полученные от модели Catboost:
Рис.2.11 Прогноз потребления воды
Рис.2.12. Особенности прогноза
Рис.2.13. Сравнение потребления воды по годам
Рис.2.14. Тепловая карта потребления воды
Рис.2.15. Среднее потребление по сезонам
Рис.2.16. Корреляция важности признаков
По характеристике предсказания модель CatBoost оказалась наиболее перспективной и удобной в работе с данными, что связано с особенностями работы самой модели и её восприятием тренировочной и тестовой выборок.
2.3 Разработка и тестирование приложения
Приложение построено по модульному принципу, что обеспечивает удобство эксплуатации кода. Графический интерфейс был построен на PyQt5 и стилизован под систему 1С из-за её лаконичности и простоты реализации. Приложение использует QTabWidget для организации контента на три основные вкладки: данные для загрузки и просмотра исходных данных, прогноз для построения графиков и анализа, и 3D анализ для расширенной визуализации. Каждая вкладка содержит специализированные виджеты, оптимизированные под конкретные задачи. Разработаны кастомные элементы управления, включая анимированные кнопки с использованием QPropertyAnimation для плавных переходов.
Для загрузчика данных я использовал алгоритм определения колонок в загружаемых файлах, поддерживающий несколько языков и различные форматы названий колонок. Алгоритм автоматически распознает столбцы с датами, потреблением и доверительными интервалами, что делает приложение гибким к различным источникам данных. Для работы с временными рядами реализованы функции автоматического преобразования дат, сортировки по времени, интерполяции пропущенных значений и расчета статистических показателей. Система поддерживает форматы CSV и Excel, используя библиотеки pandas для эффективной обработки больших объемов данных. Реализована прогрессивная загрузка с отображением прогресс-бара, что улучшает пользовательский опыт при работе с большими файлами. Данные после загрузки проходят валидацию и очистку, включая проверку на пропущенные значения и аномалии. 
Создана система аутентификации с хранением данных в JSON-файле, включающая регистрацию новых пользователей и проверку учетных данных при входе. Использование QSettings позволяет сохранять пользовательские предпочтения, включая запоминание пользователя, сохранение пути к последнему файлу и настройки интерфейса. Система обеспечивает базовый уровень безопасности через проверку введенных данных и шифрование паролей (в текущей версии пароли хранятся в открытом виде, что требует доработки для производственного использования). Интерфейс авторизации включает валидацию ввода, обработку ошибок и возможность "запомнить меня" для удобства пользователей.
Визуализация графиков строится с использованием matplotlib и включает линию основного прогноза, доверительные интервалы (если доступны), аннотации экстремальных значений, динамическую легенду и адаптивную сетку. Реализована плавная анимация построения графика с использованием интерполяции кубическими сплайнами, таймера для пошаговой отрисовки и интерактивного слайдера управления. Класс Water3DVisualizer предоставляет четыре типа 3D визуализаций: Waterfall Chart - столбчатую диаграмму в 3D пространстве, Surface Plot - поверхность тренда потребления, Ribbon Chart - ленточный график с доверительными интервалами и 3D Time Series - временной ряд с дополнительными измерениями. Каждый тип визуализации оптимизирован для определенного типа анализа и предоставляет уникальные возможности для изучения данных. Для 3D графиков реализовано интерактивное вращение с автоматической анимацией и возможностью ручного управления углом обзора.
Анализатор данных вычисляет комплекс статистических показателей: базовые метрики (среднее, минимум, максимум), показатели вариативности (стандартное отклонение, коэффициент вариации), анализ трендов и сезонные паттерны. Автоматическое формирование детальных отчетов включает статистическую сводку, анализ трендов, рекомендации по управлению и прогноз на ближайший период. Реализованы алгоритмы сглаживания данных скользящим средним для выявления долгосрочных тенденций. Система анализирует динамику потребления и выдает рекомендации в зависимости от выявленных паттернов - например, при высокой сезонной вариативности предлагает подготовить систему к пиковым нагрузкам, а при стабильном потреблении рекомендует фокусироваться на энергоэффективности.
Созданы механизмы для работы с большими объемами данных: отложенная загрузка и обработка, прогресс-бары для длительных операций, оптимизированные алгоритмы визуализации и кэширование промежуточных результатов. Достигнута высокая отзывчивость интерфейса через асинхронную обработку данных, оптимизированную перерисовку графиков, эффективное использование памяти и минимизацию блокирующих операций. Для 3D визуализаций использованы эффективные алгоритмы рендеринга и оптимизированы размеры графиков для баланса между качеством и производительностью. Система динамически подстраивает детализацию графиков в зависимости от объема данных и возможностей оборудования.

3 Охрана труда и информационная безопасность
Разработка или использование web-приложения мало связана с производственными рисками и физическими травмами сотрудников, но необходимо учитывать эргономику рабочего места и создать оптимальные условия рабочего места пользователя.
Рабочее место должно быть оборудовано в соответствии с санитарно-техническими нормами: необходимое освещение, хорошая вентиляция, звукоизоляция. Монитор должен располагаться на расстоянии 50-70 см от глаз пользователя (значение может отличаться из-за индивидуальных особенностей человека), расположения монитора должно располагаться относительно источников света таким образом, чтобы не было бликов, сам же монитор должен быть оборудован антибликовыми фильтрами, если пользователь имеет проблемы с зрением, то ему необходимо носить очки с специальным напылением, в некоторых случаях контактные линзы не могут оказать должного воздействия. Кресло должно быть регулируемым, с поддержкой поясничного и шейного отделов позвоночника, позволяющими снизить нагрузку на организм и сохранить правильную осанку. Клавиатура и мышь должны располагаться так, чтобы руки находились в нейтральном положении. 
Рекомендуется соблюдать режим работы и отдыха, каждый час необходимо делать перерывы по 5-10 минут и проводить гимнастику глаз, шеи, плеч, спины и рук. Это поможет держать тело в тонусе и держать производственную продуктивность на высоком уровне. 
Вспоминаем про безопасность работы с электрооборудованием - получить разряд электричества или оказаться посреди пожара не хочет никто. Вся электроника в округе (периферия, серверы, рабочие места и т.д.) должны быть заземлены и находиться в исправном состоянии. Если системный блок, блок питания, электрощиток или другое оборудование находится под напряжением - вскрывать их запрещается, необходимо принять действия прописанные в технике безопасности при работе с электрооборудованием. Рабочие места должны быть оснащены средствами пожаротушения (порошковый или углекислотный огнетушитель), загромождение проходов к пожарным щиткам и аварийным выходам, а также использование легковоспламеняемых материалов и воды (вода и электричество нежелательное для пользователя сочетание) строго запрещено. Регулярные проверки, правильная подготовка сотрудников и тестирование оборудования сильно уменьшит риск потенциальных аварийных ситуаций.
3.1 Информационная безопасность
Безопасность использования приложения рассматривается с точки зрения предотвращения экономических потерь для крупных фирм и заботы о рядовых пользователях. Активы приложения, которые подвержены защите это сами данные включая различные экономические факторы и персональные данные клиентов. В Российской Федерации за это отвечает закон №152-ФЗ. Данные пользователей должны храниться в отдельной базе, доступ к которой будет лишь у оператора с правами администратора. Сами модели должны обучаться на агрегированных и обезличенных данных для конспирации и для избежание утечки (в идеальном случае). Исходный код приложения - коммерческая тайна, предоставлять его необходимо в исключительных случаях по особому требованию (модификации под конкретного заказчика и т.д.). Все поступающие данные должны проходить тщательную проверку на логическое сочетание, временные ошибки и корреляцию данных. Для контроля данных используемых при переобучении моделей можно использовать хеширование для их защиты. Любые изменения в данных,  в том числе различные эксперименты при обучении моделей часто вынуждают откатываться к предыдущим версиям, а это означает использование версионной системы моделей, которая позволит избежать лишних действий при работе с ними. 
API и микросервисная архитектура зачастую не всегда может стабильно работать на постоянных паролях, более сложный метод использования токенов кратковременного действия может помочь с этим, в качестве аналога можно использовать сертификаты для подключения разных компонентов между собой.

ЗАКЛЮЧЕНИЕ
При выполнении работы по разработке приложения для предсказания потребления воды в городе получен ценный опыт в исследовании выбранной темы, сборе и предобработке данных и большее понимание сложности ручного feature engineering. 
Обучение моделей и их донастройка позволила понять механизмы работы различных моделей с точки зрения математики
Разработка приложения потребовала изучения web-дизайна и использования разных подходов к реализации ключевого и дополнительного функционала. 
Разработанный проект отвечает поставленным требованиям, но обладает широкой перспективой к развитию. Улучшение визуальной части приложения и моделей сделает приложение более привлекательным для новых пользователей. 
Приложение возможно расширить до более широко направленного сервиса, добавление нового функционала позволит улучшить работу уже имеющихся функций, но и предсказывать другие данные, к примеру не только воду, но и выброс вредных веществ в атмосферу, для этого необходимо переформатировать некоторые функции приложения и создать отдельный блок моделей для избежания перемешивания данных.
Слежка за конкурентами и анализ рыночного тренда позволит приложению развиваться в соответствии с изменяющимися условиями.  
Разработанное приложение демонстрирует комплексный подход к решению задачи прогнозирования потребления воды. Сочетание современных технологий визуализации, продуманного пользовательского интерфейса и модельной архитектуры делает его хорошим инструментом для анализа временных рядов. Ключевые достижения включают интуитивный интерфейс для сложных аналитических задач, гибкую систему обработки различных форматов данных, богатый набор визуализаций для глубокого анализа, масштабируемую архитектуру для будущего развития и профессиональный подход к обработке ошибок и пользовательскому опыту. 

СПИСОК ИСПОЛЬЗОВАННЫХ ИСТОЧНИКОВ
1. Водный кодекс Российской Федерации от 03.06.2006 № 74-ФЗ.
http://pravo.gov.ru/proxy/ips/?docbody=&nd=102107756
2. Федеральный закон от 07.12.2011 № 416-ФЗ "О водоснабжении и водоотведении".
3. Федеральный закон от 27.07.2006 № 152-ФЗ "О персональных данных".
http://pravo.gov.ru/proxy/ips/?docbody=&nd=102148912
4. Федеральный закон от 26.07.2017№187-ФЗ "О безопасности критической информационной инфраструктуры".
http://pravo.gov.ru/proxy/ips/?docbody=&nd=102429583
5. Методические рекомендации по расчету потребности в водных ресурсах для городского хозяйства (Минстрой России).
https://www.minstroyrf.gov.ru/docs/
6. РД 34.21.527-95 "Методические указания по 	составлению режимных карт работы систем водоснабжения".
https://docs.cntd.ru/document/1200004714
7. Федеральная служба государственной статистики (Росстат) - данные по водопотреблению.
https://rosstat.gov.ru/
8. Единая межведомственная информационно-статистическая система (ЕМИСС).
https://www.fedstat.ru/
9. Портал открытых данных Российской Федерации.
https://data.gov.ru/
10. World Bank Data - Water supply and sanitation statistics.
https://data.worldbank.org/topic/water
11. UN Water - Global water monitoring data.
https://www.unwater.org/
12. Hyndman, R.J., & Athanasopoulos, G. (2021) "Forecasting: Principles and Practice" (3rd ed.)
https://otexts.com/fpp3/
13. Scikit-learn Documentation: Time Series Forecasting methods.
https://scikitlearn.org/stable/modules/generated/sklearn.linear_model.LinearRegression.html
14. PyQt5 Official Documentation.
https://www.learnpyqt.com/
15. Matplotlib 	Tutorials.
https://matplotlib.org/stable/tutorials/index.html
16. Pandas Time Series.	
https://pandas.pydata.org/pandas-docs/stable/user_guide/timeseries.html
17. Stack  Overflow - решения проблем программирования.
https://stackoverflow.com/
18. TensorFlow Time Series
https://www.tensorflow.org/tutorials/structured_data/time_series
```

## 1 ТЕОРЕТИЧЕСКАЯ ЧАСТЬ

### 1.1 Методы прогнозирования временных рядов в энергетике

Временные ряды в энергетике характеризуются сложной структурой, включающей тренды, сезонность и случайные колебания. Особенности энергетических данных:
- **Мультисезонность**: суточные, недельные и годовые циклы
- **Нестационарность**: изменение статистических свойств во времени
- **Внешние воздействия**: влияние погоды, праздников, экономических факторов

**Основные подходы:**
- **Классические методы**: ARIMA, экспоненциальное сглаживание
- **Машинное обучение**: деревья решений, градиентный бустинг
- **Глубокое обучение**: LSTM, Transformer-архитектуры

### 1.2 Алгоритмы машинного обучения для задач регрессии

В работе использованы три современных алгоритма:

**Random Forest** - ансамблевый метод на основе деревьев решений, устойчивый к переобучению и хорошо работающий с разнотипными данными.

**XGBoost** - градиентный бустинг с регуляризацией, обеспечивающий высокую точность и скорость работы.

**LightGBM** - оптимизированная версия градиентного бустинга с эффективной работой с большими объемами данных.

### 1.3 Метрики оценки качества прогнозов

Для комплексной оценки моделей использованы три метрики:

**MAE (Mean Absolute Error)** - средняя абсолютная ошибка, показывающая среднее отклонение прогноза от фактических значений в абсолютных единицах.

**RMSE (Root Mean Square Error)** - среднеквадратичная ошибка, более чувствительная к большим отклонениям.

**R² (Коэффициент детерминации)** - доля объясненной дисперсии, показывающая, насколько модель лучше простого среднего.

### 2.1 Анализ и предобработка исходного датасета

**Источник данных**: Набор данных "Household Power Consumption" с Kaggle, содержащий измерения потребления электроэнергии домохозяйства с интервалом в 1 минуту за период с января по июнь 2007 года.

**Характеристики датасета:**
- Объем: 260,640 записей
- Период: 6 месяцев (181 день)
- Интервал: 1 минута
- Основные признаки: Global_active_power, Voltage, Sub_metering (3 зоны)

**Предобработка данных:**
- Обработка пропусков (1.01% данных) заполнением медианными значениями
- Создание единого datetime-индекса
- Проверка временной целостности и последовательности измерений
- Обнаружение и анализ выбросов без их удаления (реальные пики потребления)

### 2.2 Создание признаков (Feature Engineering)

Разработана комплексная система признаков, включающая первоначально 100+ параметров, с последующим отбором 33 наиболее значимых признаков для финальной модели:

**Циклические признаки:**
- hour_sin, hour_cos - циклическое кодирование часа
- day_of_week_sin, day_of_week_cos - циклическое кодирование дня недели  
- month_sin, month_cos - циклическое кодирование месяца
- day_of_year_sin, day_of_year_cos - циклическое кодирование дня года

**Временные паттерны:**
- is_weekend, is_morning_peak, is_evening_peak, is_night
- is_early_morning, is_midday, is_late_evening
- is_monday, is_friday, is_sunday, is_family_time

**Лаговые переменные** (с учетом анализа автокорреляции):
- lag_1h, lag_24h, lag_168h - стратегические лаги на 1 час, 1 день и 1 неделю

**Скользящие статистики** (со сдвигом для исключения утечек):
- rolling_mean_6h, rolling_mean_24h, rolling_mean_720h
- rolling_std_6h, rolling_std_24h

**Производные и композитные признаки:**
- momentum_1h, acceleration_1h
- deviation_from_daily_norm, volatility_ratio_6h
- composite_energy_score, behavioral_energy_pattern

### 2.3 Разведочный анализ данных (EDA)

**Анализ целевой переменной:**
- Среднее потребление: 1.156 кВт
- Медиана: 0.564 кВт (сильная правосторонняя асимметрия)
- Диапазон: 0.082 - 10.670 кВт
- Коэффициент вариации: 101.7% (высокая изменчивость)

**Суточная сезонность:**
- Минимум: 0.489 кВт в 4:00
- Максимум: 2.182 кВт в 21:00  
- Утренний пик (7-9): 1.450 кВт
- Вечерний пик (18-22): 1.897 кВт
- Ночное время (0-5): 0.581 кВт

**Недельная сезонность:**
- Рабочие дни: 1.048 кВт
- Выходные дни: 1.432 кВт (+36.6%)
- Наиболее потребляющий день: воскресенье (1.580 кВт)
- Наименее потребляющий: пятница (0.938 кВт)

**Месячная сезонность:**
- Январь: 1.546 кВт (максимум)
- Июнь: 0.827 кВт (минимум) 
- Снижение зима→лето: -43.9%

**Анализ автокорреляции:**
- Сильная корреляция на лаге 1 день (+0.4915)
- Выраженная недельная сезонность (лаг 7 дней: +0.4008)
- Значимые корреляции до 14 дней

### 2.4 Построение и обучение моделей машинного обучения

**Подготовка данных:**
- Временное разделение: train (01.01.2007 - 14.04.2007) / test (29.04.2007 - 30.06.2007)
- **Временной буфер: 14 дней для исключения утечек данных**
- Финальный набор: 33 наиболее значимых признака

*Примечание: Использование временного буфера привело к исключению данных 
за период 15-28 апреля 2007 года, что является методическим ограничением 
данного исследования.*


"При временном разделении использован буфер 14 дней, однако 
лаговые признаки (lag_1h, lag_24h и др.) используют данные 
из буферной зоны, что создает временные утечки данных 
и требует дополнительной доработки метода создания признаков."


**Обучение моделей:**
- RandomForest: 100 деревьев, max_depth=20, random_state=42
- XGBoost: 100 estimators, max_depth=10, learning_rate=0.1, random_state=42  
- LightGBM: 100 estimators, max_depth=10, learning_rate=0.1, random_state=42

**Результаты сравнения:**

| Модель | MAE (кВт) | RMSE (кВт) | R² | Интерпретация |
|--------|-----------|------------|----|---------------|
| RandomForest | 0.1536 | 0.3108 | 0.8977 | Хорошее качество |
| XGBoost | 0.1392 | 0.3054 | 0.9012 | Высокое качество |
| **LightGBM** | **0.1077** | **0.2821** | **0.9157** | **Наилучшее качество** |

**LightGBM показала наилучшие результаты:**
- Средняя ошибка: 107.7 Вт
- Объясненная дисперсия: 91.57%
- Максимальная ошибка: 3.78 кВт

**Анализ важности признаков:**
Топ-5 наиболее значимых признаков:
1. lag_1h (13.23%) - потребление час назад
2. momentum_1h (11.87%) - тренд изменения
3. acceleration_1h (9.13%) - ускорение изменения
4. rolling_std_6h (6.57%) - волатильность
5. rolling_mean_6h (5.80%) - краткосрочное среднее

**Распределение по категориям признаков:**
- Временные лаги: 39.5%
- Скользящие статистики: 26.8% 
- Производные признаки: 17.1%
- Циклические: 6.2%
- Сезонность: 6.1%

**Размеры выборок:**
- Обучающая выборка: 149,760 записей (57.4% от общего объема)
- Тестовая выборка: 90,720 записей (34.8% от общего объема)  
- Временной буфер: 14 дней (7.8% данных исключено для чистоты эксперимента)


**Преимущества LightGBM для временных рядов:**
- Эффективная работа с категориальными признаками (временные метки)
- Устойчивость к переобучению благодаря регуляризации
- Оптимизированные вычисления для больших объемов данных

### 2.6 Анализ результатов и выявление систематических ошибок

**Выявленные особенности работы модели:**
1. "Значительное сезонное смещение - различие в среднем потреблении между train (1.360 кВт) и test (0.898 кВт) выборками составляет 34%" 
2. **Ограниченность данных** - обучение только на 6 месяцах 2007 года
3. **Сезонное смещение** - различие в среднем потреблении между train (1.360 кВт) и test (0.898 кВт) выборками

**Анализ качества прогнозов:**
- Модель корректно улавливает временные паттерны и суточные циклы
- Сохранение правильной формы суточного профиля с пиками в ожидаемое время
- Разные прогнозы для разных дней (1.11 кВт vs 0.96 кВт), что подтверждает обучение на временных зависимостях
- Основное достижение: сохранение относительных соотношений между периодами потребления

**Технические проблемы:**
- Большой временной разрыв между train и test выборками (14 дней)
- Необходимость генерации признаков вместо использования реальных исторических данных
- Ограничения, связанные с данными только за 2007 год

**Предложения по улучшению:**
1. Сбор данных за полный годовой цикл для учета всех сезонных изменений
2. Использование walk-forward validation вместо единого разделения
3. Добавление внешних факторов (погодные условия, праздничные дни)
4. Реализация онлайн-обучения для адаптации к изменяющимся паттернам потребления
5. Эксперименты с специализированными архитектурами для временных рядов (LSTM, Transformer)

        	Информированное потребление. Наглядная визуализация прогноза на следующие сутки помогает пользователям лучше понимать свои паттерны потребления и принимать осознанные решения.

1 АНАЛИЗ ПРЕДМЕТНОЙ ОБЛАСТИ И ПОСТАНОВКА ЗАДАЧИ НА ПРОЕКТИРОВАНИЕ

	Объектом для проектирования информационной системы прогнозирования, рассматриваемым в данной работе, является процесс суточного энергопотребления домохозяйств. Для проектирования системы был взят за основу подход к решению задач временных рядов с использованием машинного обучения. На рисунке 1 представлена общая архитектура подобных систем прогнозирования.


Рисунок 1 – Общая архитектура системы прогнозирования временных рядов
	В данной системе предусмотрена функциональность анализа исторических данных, выявления сезонных паттернов и генерации признаков для модели машинного обучения. Также система включает модуль для интерактивного взаимодействия с пользователем.
	В работе системы участвуют два основных типа данных: исторические показания потребления электроэнергии и сгенерированные временные признаки. Исторические данные включают значения потребления с временными метками, которые используются для выявления закономерностей и обучения модели. На основе этих данных формируются признаки, такие как время суток, день недели, скользящие средние и лаговые значения, которые позволяют модели делать точные прогнозы.
	Задача информационной системы – автоматизация процесса прогнозирования энергопотребления, что позволяет пользователям заранее планировать свою активность и оптимизировать затраты на электроэнергию.
	Таким образом, предметом автоматизации является создание информационной системы, позволяющей на основе исторических данных строить точные прогнозы потребления электроэнергии на ближайшие сутки, а также предоставляющей удобный интерфейс для получения этих прогнозов.
	Проведенный анализ предметной области позволяет сформулировать задачу на проектирование:
 	разработать систему прогнозирования суточного потребления электроэнергии с использованием алгоритмов машинного обучения
 	реализовать Telegram-бота для интерактивного предоставления прогнозов пользователям
	Решение задачи с использованием современных информационных технологий и библиотек машинного обучения представляется возможным.
	Поставленную задачу на проектирование можно разделить на подзадачи, требующие конкретной реализации.
	Необходимо реализовать:
 	проведение анализа и предобработки данных энергопотребления
 	выявление сезонных закономерностей и временных паттернов
 	создание комплексной системы признаков для модели машинного обучения
 	обучение и сравнение нескольких алгоритмов прогнозирования
 	разработку Telegram-бота для интерактивного тестирования модели
 	анализ качества прогнозов и выявление систематических ошибок
	Экономически перспективной задача по разработке данной информационной системы выглядит по причине того, что система позволяет оптимизировать энергопотребление, снизить затраты пользователей на электроэнергию, а также может быть масштабирована для использования энергосбытовыми компаниями для балансировки нагрузок в сети.

 	В качестве исходных данных для построения системы прогнозирования использован датасет "Household Power Consumption", содержащий детальные измерения потребления электроэнергии с интервалом в одну минуту. Период измерений охватывает шесть месяцев с января по июнь 2007 года.
 	Исходный датасет характеризуется следующими параметрами: общий объем данных составляет 260 640 записей, временной период охватывает 181 день, частота измерений равна одной минуте. Основные показатели включают глобальную активную мощность, напряжение и потребление по трем зонам учета.

	На этапе предобработки данных решены следующие задачи: выявление и обработка пропущенных значений, составляющих 1.01% от общего объема данных; создание единого временного индекса для обеспечения целостности временного ряда; проверка последовательности измерений и устранение временных разрывов; анализ и верификация выбросов без их удаления, поскольку экстремальные значения представляют реальные пики энергопотребления.

 	На основании комплексного анализа метрик качества и интерпретируемости модели для финальной реализации выбрана модель LightGBM, показавшая наилучшие результаты прогнозирования суточного потребления электроэнергии.
 	Для обеспечения удобного доступа к функционалу прогнозирования разработан Telegram-бот, предоставляющий пользовательский интерфейс для взаимодействия с обученной моделью LightGBM. Выбор платформы Telegram обусловлен ее распространенностью, кроссплатформенностью и минимальными требованиями к инфраструктуре.
 	Архитектура бота реализована по модульному принципу и включает следующие компоненты: модуль конфигурации (config.py), модуль загрузки данных (data_loader.py), модуль генерации признаков (feature_generator.py) и основной модуль бота (bot.py). Для разработки использован фреймворк python-telegram-bot версии 13.7.
Бот поддерживает следующий набор команд и функций:
 	команда /start обеспечивает начальное взаимодействие с пользователем, выводя приветственное сообщение и описание возможностей системы
 	команда /predict предоставляет доступ к меню прогнозирования с интерактивными кнопками для выбора временного горизонта
 	команда /stats выводит информацию о метриках качества модели и используемых алгоритмах
 	система инлайн-кнопок позволяет пользователю выбирать между прогнозами на завтра и послезавтра, а также сравнивать оба варианта
 	Интерфейс бота разработан с учетом принципов юзабилити и обеспечивает интуитивно понятное взаимодействие даже для неподготовленных пользователей. Визуальное представление результатов включает графики прогнозируемого потребления с выделением ключевых характеристик временного ряда.
 	Ключевым аспектом интеграции является обеспечение корректного взаимодействия между Telegram-ботом и обученной моделью LightGBM. Для этого разработан специализированный модуль генерации признаков, который создает входные данные для модели на основе выявленных в ходе EDA паттернов энергопотребления.
 	При получении запроса на прогноз от пользователя система выполняет следующую последовательность действий. Модуль feature_generator.py генерирует полный набор из 33 признаков для запрашиваемой даты, включая циклические кодирования времени, лаговые переменные и скользящие статистики. Особенностью реализации является использование синтетических данных, основанных на статистических закономерностях исходного датасета, что обусловлено демонстрационным характером прототипа.
 	Сгенерированные признаки передаются в обученную модель LightGBM, которая формирует прогноз потребления электроэнергии с часовым разрешением на следующие 24 часа. Результат прогнозирования включает не только точечные оценки, но и визуализацию в виде графика, отображающего ожидаемую суточную динамику потребления.
 	Для обеспечения реалистичности прогнозов в алгоритм генерации признаков заложены выявленные ранее закономерности: сезонные коэффициенты (увеличение потребления на 33.5% в январе и снижение на 28.5% в июне), недельные паттерны (повышение потребления на 36.6% в выходные дни) и корректное отражение суточной динамики с ночными минимумами и вечерними пиками.
 	Система демонстрирует устойчивую работу при тестировании на различных датах, показывая различные результаты прогнозирования (1.11 кВт против 0.96 кВт для разных дней) при сохранении характерной формы суточного профиля потребления.
 	Для верификации корректности работы системы проведено комплексное тестирование прототипа, включающее проверку функциональности Telegram-бота и точности формируемых прогнозов.
 	Тестирование пользовательского интерфейса подтвердило работоспособность всех команд бота. Команда /start корректно выводит приветственное сообщение и описание возможностей системы. Команда /predict обеспечивает отображение интерактивных кнопок для выбора временного горизонта прогнозирования. Команда /stats возвращает актуальные метрики качества обученной модели.
 	Проверка системы прогнозирования на тестовых сценариях показала, что модель сохраняет правильную форму суточного профиля потребления с характерными пиками в утренние и вечерние часы. При тестировании на различные даты система демонстрирует различные результаты прогнозирования, что подтверждает ее способность учитывать временные зависимости.
 	Функциональность визуализации работает корректно - для каждого прогноза формируется график с четким отображением ожидаемой динамики потребления. Система стабильно обрабатывает запросы и не проявляет признаков аномального поведения при длительной работе.
 	Интеграция между компонентами системы (модуль генерации признаков, модель машинного обучения и Telegram-бот) функционирует без сбоев, обеспечивая полный цикл обработки запроса от пользователя до формирования итогового прогноза.
	В ходе тестирования системы выявлен ряд систематических ошибок и ограничений, оказывающих влияние на качество прогнозирования. Основной проблемой является значительное сезонное смещение между обучающей и тестовой выборками - среднее потребление в train-выборке составляет 1.360 кВт, тогда как в test-выборке этот показатель равен 0.898 кВт, что соответствует разнице 34%.
 	Другим существенным ограничением является временной разрыв между выборками продолжительностью 14 дней, обусловленный методикой создания лаговых признаков. Этот разрыв приводит к потере репрезентативности данных и может негативно сказываться на качестве прогнозов для периодов со значительными сезонными изменениями.
 	В текущей реализации Telegram-бота используется генерация признаков на основе статистических паттернов EDA-анализа, а не реальных исторических данных, что является методическим ограничением для точного прогнозирования в реальных условиях. Кроме того, заявленные метрики качества модели были достигнуты в условиях лабораторного тестирования и могут отличаться при работе с реальными поступающими данными.
 	Для устранения выявленных ограничений предложены следующие направления развития системы: сбор данных за полный годовой цикл для учета всех сезонных изменений; использование walk-forward validation вместо единого временного разделения; добавление внешних факторов влияния, таких как погодные условия и календарь праздников; реализация механизма онлайн-обучения для адаптации модели к изменяющимся паттернам потребления.
 	Несмотря на выявленные систематические ошибки, разработанная система демонстрирует работоспособность и может служить основой для создания более совершенных решений в области прогнозирования энергопотребления.

Учитывая демонстрационный характер прототипа, основные риски безопасности связаны с обработкой данных и взаимодействием с внешними сервисами.

 	Для потенциальной промышленной эксплуатации системы рекомендовано  реализовать дополнительные меры безопасности: внедрение системы мониторинга аномальной активности, регулярное обновление используемых программных компонентов, проведение пентест-испытаний и разработка политики резервного копирования данных.

Взаимодействие с системой происходит через защищенное соединение, обеспечивающее конфиденциальность передаваемых данных.
Эргономические аспекты
 	Интерфейс Telegram-бота разработан с учетом эргономических требований: используется достаточный контраст текста, исключены мерцающие элементы, обеспечена логичная навигация. Это минимизирует зрительную нагрузку и предотвращает потенциальные ошибки при взаимодействии с системой.

 	В процессе работы решен комплекс задач, включающий анализ и предобработку данных энергопотребления, выявление сезонных закономерностей и временных паттернов, создание системы признаков, обучение и сравнение алгоритмов машинного обучения.
